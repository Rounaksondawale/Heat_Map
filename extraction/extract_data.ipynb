```python
import os

file = "pune_era5_2023_02_06_08.grib"

print("Current folder:")
print(os.getcwd())

print("\nFile exists:")
print(os.path.exists(file))

if os.path.exists(file):
    print("\nFile size:")
    print(round(os.path.getsize(file) / (1024 * 1024), 2), "MB")
```

```python
import xarray as xr

ds = xr.open_dataset(
    "pune_era5_2023_02_06_08.grib",
    engine="cfgrib",
    backend_kwargs={
        "indexpath": ""
    }
)

print(ds)
```

```python
print("Dimensions:")

for dimension, size in ds.sizes.items():
    print(f"{dimension}: {size}")
```
